```python
import plotly.express as px

matches = [1, 2, 3, 4, 5, 6]
runs = [45, 78, 102, 56, 89, 120]

fig = px.line(x=matches, y=runs, markers=True,
              title="Cricket Player Runs per Match",
              labels={"x": "Match Number", "y": "Runs Scored"})

fig.show()
```

```python
import plotly.express as px

teams = ["India", "Australia", "England", "Pakistan", "South Africa"]
total_runs = [320, 280, 305, 290, 275]

fig = px.bar(x=teams, y=total_runs,
             title="Team Total Runs Comparison",
             labels={"x": "Team", "y": "Total Runs"},
             color=total_runs)

fig.show()
```

```python
import plotly.express as px

runs = [45, 78, 102, 56, 89, 120]
strike_rate = [110, 125, 140, 115, 130, 150]

fig = px.scatter(x=runs, y=strike_rate,
                 title="Runs vs Strike Rate",
                 labels={"x": "Runs", "y": "Strike Rate"},
                 trendline="ols")

fig.show()
```

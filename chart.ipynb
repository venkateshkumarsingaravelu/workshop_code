```python
import matplotlib.pyplot as plt

# Sample cricket data
matches = [1, 2, 3, 4, 5, 6]
runs = [45, 78, 102, 56, 89, 120]

plt.figure(figsize=(10, 5))
plt.plot(matches, runs, marker='o', linestyle='-', color='blue')

plt.title("Cricket Player Performance")
plt.xlabel("Match Number")
plt.ylabel("Runs Scored")
# plt.grid(axis ='x')
plt.grid(color = 'green', linestyle = '--', linewidth = 0.5)
# Save instead of show (works even if Tkinter is broken)
plt.savefig("cricket_scores.png")

print("Plot saved as cricket_scores.png")
```
